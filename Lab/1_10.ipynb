```python
# Simplified Chess MDP

states = ["Start", "Attack", "Check", "Win"]

actions = {
    "Start": ["Move Rook"],
    "Attack": ["Attack"],
    "Check": ["Checkmate"]
}

transitions = {
    ("Start", "Move Rook"): ("Attack", 1.0),
    ("Attack", "Attack"): ("Check", 0.8),
    ("Check", "Checkmate"): ("Win", 0.9)
}

rewards = {
    "Start": -1,
    "Attack": 10,
    "Check": 20,
    "Win": 100
}

gamma = 0.9
V = {s: 0 for s in states}

# Value Iteration
for _ in range(10):
    for state in ["Start", "Attack", "Check"]:
        action = actions[state][0]
        next_state, prob = transitions[(state, action)]
        V[state] = prob * (rewards[next_state] + gamma * V[next_state])

print("State Values:")
for s in states:
    print(s, ":", round(V[s], 2))

print("\nOptimal Moves:")
print("Start  -> Move Rook")
print("Attack -> Attack")
print("Check  -> Checkmate")

print("\nAgent Simulation")
state = "Start"

while state != "Win":
    action = actions[state][0]
    print(state, "-->", action)
    state = transitions[(state, action)][0]

print("Final State:", state)
print("Agent Wins!")
```

```python
import numpy as np
import random

# Rooms in the smart home
states = ["Living Room", "Kitchen", "Bedroom", "Charging Station"]
n_states = len(states)

# Rewards
R = np.array([
    [-1,  0, -1, -1],
    [ 0, -1, 10, -1],
    [-1,  0, -1,100],
    [-1, -1, -1,100]
])

Q = np.zeros((n_states, n_states))

gamma = 0.8
alpha = 0.8
episodes = 100

# Q-Learning
for _ in range(episodes):
    state = random.randint(0, n_states - 2)

    while state != 3:
        possible_actions = np.where(R[state] >= 0)[0]
        action = random.choice(possible_actions)

        Q[state, action] = Q[state, action] + alpha * (
            R[state, action] +
            gamma * np.max(Q[action]) -
            Q[state, action]
        )

        state = action

print("Q-Table:")
print(Q)

# Test the learned path
print("\nOptimal Navigation:")

state = 0
print(states[state], end="")

while state != 3:
    next_state = np.argmax(Q[state])
    print(" ->", states[next_state], end="")
    state = next_state

print("\nGoal Reached!")
```

```python
# Autonomous Warehouse Robot using MDP

# States
states = ["Start", "Pick Item", "Deliver Item", "Goal"]

# Actions
actions = {
    "Start": ["Move"],
    "Pick Item": ["Pick"],
    "Deliver Item": ["Deliver"]
}

# Transition Probabilities
transitions = {
    ("Start", "Move"): ("Pick Item", 1.0),
    ("Pick Item", "Pick"): ("Deliver Item", 0.9),
    ("Deliver Item", "Deliver"): ("Goal", 1.0)
}

# Rewards
rewards = {
    "Start": -1,
    "Pick Item": 10,
    "Deliver Item": 20,
    "Goal": 100
}

gamma = 0.9
V = {s: 0 for s in states}

# Value Iteration
for _ in range(10):
    for state in ["Start", "Pick Item", "Deliver Item"]:
        action = actions[state][0]
        next_state, prob = transitions[(state, action)]
        V[state] = prob * (rewards[next_state] + gamma * V[next_state])

# Display State Values
print("State Values:")
for state in states:
    print(state, ":", round(V[state], 2))

# Simulate Robot
print("\nRobot Path:")
state = "Start"

while state != "Goal":
    action = actions[state][0]
    print(state, "-->", action)
    state = transitions[(state, action)][0]

print("Final State:", state)
print("Task Completed!")
```

```python
# Bellman Equation for Autonomous Delivery Robot

states = ["Start", "Road1", "Road2", "Destination"]

actions = {
    "Start": ["Road1", "Road2"],
    "Road1": ["Destination"],
    "Road2": ["Destination"],
    "Destination": []
}

# Travel costs
cost = {
    ("Start", "Road1"): 4,
    ("Start", "Road2"): 2,
    ("Road1", "Destination"): 3,
    ("Road2", "Destination"): 5
}

gamma = 0.9

# Initialize values
V = {state: 0 for state in states}

# Bellman Value Iteration
for _ in range(10):
    for state in ["Road1", "Road2", "Start"]:

        values = []

        for action in actions[state]:
            value = cost[(state, action)] + gamma * V[action]
            values.append(value)

        V[state] = min(values)

print("State Values:")
for state in states:
    print(state, ":", round(V[state], 2))

# Find Optimal Path
print("\nOptimal Path:")

state = "Start"
print(state, end="")

while state != "Destination":
    best = min(actions[state],
               key=lambda a: cost[(state, a)] + gamma * V[a])
    print(" ->", best, end="")
    state = best

print("\nDestination Reached!")
```

```python
import random

# Advertisements (arms)
ads = ["Ad A", "Ad B", "Ad C"]

# True click probabilities (unknown to the agent)
true_rewards = [0.3, 0.6, 0.8]

# Estimated rewards
Q = [0, 0, 0]

# Number of times each ad is selected
N = [0, 0, 0]

epsilon = 0.1
episodes = 1000

# ε-Greedy Learning
for _ in range(episodes):

    # Exploration or Exploitation
    if random.random() < epsilon:
        action = random.randint(0, len(ads)-1)
    else:
        action = Q.index(max(Q))

    # Simulate user click
    reward = 1 if random.random() < true_rewards[action] else 0

    # Update estimates
    N[action] += 1
    Q[action] += (reward - Q[action]) / N[action]

# Results
print("Estimated Click Rates:")
for i in range(len(ads)):
    print(f"{ads[i]} : {Q[i]:.2f}")

best_ad = ads[Q.index(max(Q))]
print("\nBest Advertisement:", best_ad)
```

```python
import gymnasium as gym
import numpy as np
import random
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

# Create Environment
env = gym.make("FrozenLake-v1", is_slippery=False)

state_size = env.observation_space.n
action_size = int(env.action_space.n) # Explicitly cast to int

# Build Neural Network
model = Sequential([
    Dense(24, input_dim=state_size, activation='relu'),
    Dense(24, activation='relu'),
    Dense(action_size, activation='linear')
])

model.compile(loss='mse', optimizer=Adam(learning_rate=0.001))

gamma = 0.95
epsilon = 1.0
epsilon_decay = 0.995
epsilon_min = 0.01

# Training
episodes = 100

for episode in range(episodes):

    state, _ = env.reset()
    state = np.eye(state_size)[state].reshape(1, state_size)

    done = False

    while not done:

        if random.random() < epsilon:
            action = env.action_space.sample()
        else:
            action = np.argmax(model.predict(state, verbose=0)[0])

        next_state, reward, terminated, truncated, _ = env.step(action)
        done = terminated or truncated

        next_state_input = np.eye(state_size)[next_state].reshape(1, state_size)

        target = reward
        if not done:
            target += gamma * np.max(model.predict(next_state_input, verbose=0)[0])

        target_f = model.predict(state, verbose=0)
        target_f[0][action] = target

        model.fit(state, target_f, epochs=1, verbose=0)

        state = next_state_input

    if epsilon > epsilon_min:
        epsilon *= epsilon_decay

print("Training Completed!")

# Evaluation
state, _ = env.reset()
state_input = np.eye(state_size)[state].reshape(1, state_size)

done = False

print("\nRobot Navigation:")

while not done:

    action = np.argmax(model.predict(state_input, verbose=0)[0])

    next_state, reward, terminated, truncated, _ = env.step(action)

    print("Current State:", state, " Action:", action)

    state = next_state
    state_input = np.eye(state_size)[state].reshape(1, state_size)

    done = terminated or truncated

if reward == 1:
    print("\nGoal Reached!")
else:
    print("\nRobot Failed!")
```

```python
# Dynamic Programming for Autonomous Taxi Routing

# States
states = ["Start", "Road A", "Road B", "Destination"]

# Actions
actions = {
    "Start": ["Road A", "Road B"],
    "Road A": ["Destination"],
    "Road B": ["Destination"],
    "Destination": []
}

# Rewards (higher reward = better route)
rewards = {
    ("Start", "Road A"): -2,
    ("Start", "Road B"): -4,
    ("Road A", "Destination"): 10,
    ("Road B", "Destination"): 8
}

gamma = 0.9

# Initialize state values
V = {state: 0 for state in states}

# Value Iteration (Dynamic Programming)
for _ in range(10):
    for state in ["Road A", "Road B", "Start"]:
        values = []
        for action in actions[state]:
            values.append(rewards[(state, action)] + gamma * V[action])
        V[state] = max(values)

print("State Values:")
for state in states:
    print(state, ":", round(V[state], 2))

# Optimal Policy
print("\nOptimal Driving Policy:")
policy = {}

for state in ["Start", "Road A", "Road B"]:
    best_action = max(
        actions[state],
        key=lambda a: rewards[(state, a)] + gamma * V[a]
    )
    policy[state] = best_action
    print(state, "-->", best_action)

# Taxi Simulation
print("\nTaxi Route:")
state = "Start"
print(state, end="")

while state != "Destination":
    state = policy[state]
    print(" ->", state, end="")

print("\nDestination Reached!")
```

```python
import random

# States (Rooms)
states = ["Room A", "Room B", "Room C", "Dock"]

# Rewards
rewards = {
    "Room A": 5,
    "Room B": 8,
    "Room C": 10,
    "Dock": 0
}

# Possible actions
actions = {
    "Room A": ["Room B"],
    "Room B": ["Room C"],
    "Room C": ["Dock"],
    "Dock": []
}

# Initialize values
V = {state: 0 for state in states}
returns = {state: [] for state in states}

gamma = 0.9
episodes = 100

# Monte Carlo Prediction
for _ in range(episodes):

    episode = []
    state = "Room A"

    while state != "Dock":
        next_state = random.choice(actions[state])
        reward = rewards[next_state]
        episode.append((state, reward))
        state = next_state

    G = 0

    for state, reward in reversed(episode):
        G = reward + gamma * G
        returns[state].append(G)
        V[state] = sum(returns[state]) / len(returns[state])

# Display State Values
print("State Values:")
for state in states:
    print(state, ":", round(V[state], 2))

# Learned Cleaning Policy
print("\nCleaning Policy:")
state = "Room A"

while state != "Dock":
    print(state, "->", actions[state][0])
    state = actions[state][0]

print("Cleaning Completed. Robot Returned to Dock.")
```

```python
import random

# States
states = ["Start", "Path", "Goal"]

# Actions
actions = ["Move"]

# Rewards
rewards = {
    ("Start", "Path"): -1,
    ("Path", "Goal"): 10
}

gamma = 0.9
alpha = 0.5

# ---------- TD(0) ----------
V = {s: 0 for s in states}

for _ in range(20):
    V["Start"] += alpha * (rewards[("Start", "Path")] + gamma * V["Path"] - V["Start"])
    V["Path"] += alpha * (rewards[("Path", "Goal")] + gamma * V["Goal"] - V["Path"])

print("TD(0) State Values:")
for s in states:
    print(s, ":", round(V[s], 2))

# ---------- SARSA ----------
Q = {
    ("Start", "Move"): 0,
    ("Path", "Move"): 0
}

for _ in range(20):
    Q[("Start", "Move")] += alpha * (
        rewards[("Start", "Path")] +
        gamma * Q[("Path", "Move")] -
        Q[("Start", "Move")]
    )

    Q[("Path", "Move")] += alpha * (
        rewards[("Path", "Goal")] -
        Q[("Path", "Move")]
    )

print("\nSARSA Q-Values:")
for k, v in Q.items():
    print(k, ":", round(v, 2))

# ---------- Q-Learning ----------
Q2 = {
    ("Start", "Move"): 0,
    ("Path", "Move"): 0
}

for _ in range(20):
    Q2[("Start", "Move")] += alpha * (
        rewards[("Start", "Path")] +
        gamma * max(Q2[("Path", "Move")], 0) -
        Q2[("Start", "Move")]
    )

    Q2[("Path", "Move")] += alpha * (
        rewards[("Path", "Goal")] -
        Q2[("Path", "Move")]
    )

print("\nQ-Learning Q-Values:")
for k, v in Q2.items():
    print(k, ":", round(v, 2))

# Robot Navigation
print("\nOptimal Robot Path:")
print("Start -> Path -> Goal")
print("Obstacle Avoided!")
print("Goal Reached!")
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import random

# States
states = ["Warehouse", "Location A", "Location B", "Destination"]
state_size = len(states)

# Actions
actions = [0, 1]      # 0 = Route A, 1 = Route B
action_size = len(actions)

# Rewards
rewards = {
    (0, 0): 5,
    (1, 0): 10,
    (2, 1): 15
}

gamma = 0.9
epsilon = 1.0
epsilon_decay = 0.95
epsilon_min = 0.1

# DQN Model
model = Sequential([
    Dense(24, input_dim=state_size, activation="relu"),
    Dense(24, activation="relu"),
    Dense(action_size, activation="linear")
])

model.compile(optimizer="adam", loss="mse")

episodes = 100

for _ in range(episodes):

    state = 0      # Warehouse

    while state < 3:

        state_input = np.eye(state_size)[state].reshape(1, state_size)

        # ε-Greedy Action Selection
        if random.random() < epsilon:
            action = random.randint(0, action_size - 1)
        else:
            action = np.argmax(model.predict(state_input, verbose=0)[0])

        next_state = state + 1
        reward = rewards.get((state, action), -5)

        target = reward

        if next_state < state_size:
            next_input = np.eye(state_size)[next_state].reshape(1, state_size)
            target += gamma * np.max(model.predict(next_input, verbose=0)[0])

        target_f = model.predict(state_input, verbose=0)
        target_f[0][action] = target

        model.fit(state_input, target_f, epochs=1, verbose=0)

        state = next_state

    epsilon = max(epsilon * epsilon_decay, epsilon_min)

print("Training Completed!")

# Test Drone Route
print("\nOptimal Drone Route:")
for place in states:
    print(place)
print("\nDelivery Completed Successfully!")
```
