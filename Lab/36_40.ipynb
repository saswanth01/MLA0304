```python
import random

Q = {
    "collect": [0, 0],
    "process": [0, 0],
    "deliver": [0, 0]
}

tasks = ["collect", "process", "deliver"]

for episode in range(100):
    total = 0

    for task in tasks:
        action = random.randint(0, 1)

        reward = 1 if action == 1 else 0
        Q[task][action] += 0.1 * (reward - Q[task][action])
        total += reward

print("Learned MAXQ Policies:")

for task in tasks:
    print(task, "-> Action", Q[task].index(max(Q[task])))

print("Final Task Score:", total, "/ 3")
```

```python
import random

states = ["A", "B", "C", "D", "Goal"]

def navigate():
    state = "A"
    path = [state]

    while state != "Goal":
        observation = random.choice(["clear", "blocked"])

        if state == "A":
            state = "B" if observation == "clear" else "C"
        elif state in ["B", "C"]:
            state = "D"
        elif state == "D":
            state = "Goal"

        path.append(state)

    return path

success = 0

for i in range(5):
    path = navigate()
    if path[-1] == "Goal":
        success += 1
    print("Episode", i+1, ":", path)

print("Success Rate:", success / 5 * 100, "%")
```

```python
import random

Q = [0, 0]
patients = 10

for episode in range(100):
    total_reward = 0

    for i in range(patients):
        action = random.randint(0, 1)

        if action == 0:
            reward = 10
        else:
            reward = -2

        Q[action] += 0.1 * (reward - Q[action])
        total_reward += reward

print("Healthcare RL Policy:")
print("Best Action:", Q.index(max(Q)))
print("Q Values:", [round(x, 2) for x in Q])
print("Final Reward:", total_reward)
```

```python
import random

Q = [0, 0, 0]
levels = ["Easy", "Medium", "Hard"]

for episode in range(100):
    ability = random.randint(0, 2)

    action = random.randint(0, 2)

    reward = 1 if action == ability else -1

    Q[action] += 0.1 * (reward - Q[action])

print("Personalized Learning Policy:")

for i in range(3):
    print(levels[i], "Q =", round(Q[i], 2))

print("Recommended Level:",
      levels[Q.index(max(Q))])
```
