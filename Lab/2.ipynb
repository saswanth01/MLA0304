```python
import numpy as np
import random

# Rooms in the smart home
states = ["Living Room", "Kitchen", "Bedroom", "Charging Station"]
n_states = len(states)

# Rewards
R = np.array([
    [-1,  0, -1, -1],
    [ 0, -1, 10, -1],
    [-1,  0, -1,100],
    [-1, -1, -1,100]
])

Q = np.zeros((n_states, n_states))

gamma = 0.8
alpha = 0.8
episodes = 100

# Q-Learning
for _ in range(episodes):
    state = random.randint(0, n_states - 2)

    while state != 3:
        possible_actions = np.where(R[state] >= 0)[0]
        action = random.choice(possible_actions)

        Q[state, action] = Q[state, action] + alpha * (
            R[state, action] +
            gamma * np.max(Q[action]) -
            Q[state, action]
        )

        state = action

print("Q-Table:")
print(Q)

# Test the learned path
print("\nOptimal Navigation:")

state = 0
print(states[state], end="")

while state != 3:
    next_state = np.argmax(Q[state])
    print(" ->", states[next_state], end="")
    state = next_state

print("\nGoal Reached!")
```
