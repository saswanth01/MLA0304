```python
# Simplified Chess MDP

states = ["Start", "Attack", "Check", "Win"]

actions = {
    "Start": ["Move Rook"],
    "Attack": ["Attack"],
    "Check": ["Checkmate"]
}

transitions = {
    ("Start", "Move Rook"): ("Attack", 1.0),
    ("Attack", "Attack"): ("Check", 0.8),
    ("Check", "Checkmate"): ("Win", 0.9)
}

rewards = {
    "Start": -1,
    "Attack": 10,
    "Check": 20,
    "Win": 100
}

gamma = 0.9
V = {s: 0 for s in states}

# Value Iteration
for _ in range(10):
    for state in ["Start", "Attack", "Check"]:
        action = actions[state][0]
        next_state, prob = transitions[(state, action)]
        V[state] = prob * (rewards[next_state] + gamma * V[next_state])

print("State Values:")
for s in states:
    print(s, ":", round(V[s], 2))

print("\nOptimal Moves:")
print("Start  -> Move Rook")
print("Attack -> Attack")
print("Check  -> Checkmate")

print("\nAgent Simulation")
state = "Start"

while state != "Win":
    action = actions[state][0]
    print(state, "-->", action)
    state = transitions[(state, action)][0]

print("Final State:", state)
print("Agent Wins!")
```
